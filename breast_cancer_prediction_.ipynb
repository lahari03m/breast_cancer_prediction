```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
```

```
data.data                 # the 30 feature values for every patient
```

```
print(data.DESCR)         # full description of the dataset
```

```
df = pd.DataFrame(data.data, columns=data.feature_names)
```

```
df['target'] = data.target
```

```
x = data.data
y = data.target
```

```
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```
x_train.shape, x_test.shape
```

```
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

model1 = LogisticRegression()
model1.fit(x_train, y_train)
y_pred1 = model1.predict(x_test)
acc1 = accuracy_score(y_test, y_pred1)
acc1
```

```
from sklearn.neighbors import KNeighborsClassifier
model2 = KNeighborsClassifier()
model2.fit(x_train, y_train)
y_pred2 = model2.predict(x_test)
acc2 = accuracy_score(y_test, y_pred2)
acc2
```

```
from sklearn.svm import SVC
model3 = SVC()
model3.fit(x_train, y_train)
y_pred3 = model3.predict(x_test)
acc3 = accuracy_score(y_test, y_pred3)
acc3
```

```
from sklearn.naive_bayes import GaussianNB
model4 = GaussianNB()
model4.fit(x_train, y_train)
y_pred4 = model4.predict(x_test)
acc4 = accuracy_score(y_test, y_pred4)
acc4
```

```
from sklearn.tree import DecisionTreeClassifier
model5 = DecisionTreeClassifier()
model5.fit(x_train, y_train)
y_pred5 = model5.predict(x_test)
acc5 = accuracy_score(y_test, y_pred5)
acc5
```

```
from sklearn.ensemble import RandomForestClassifier
model6 = RandomForestClassifier()
model6.fit(x_train, y_train)
y_pred6 = model6.predict(x_test)
acc6 = accuracy_score(y_test, y_pred6)
acc6
```

```
results = pd.DataFrame({
    'Algorithm': ['Logistic Regression', 'KNN', 'SVM', 'Naive Bayes', 'Decision Tree', 'Random Forest'],
    'Accuracy': [acc1, acc2, acc3, acc4, acc5, acc6]
})
results
results.sort_values('Accuracy', ascending=False)
```

```
plt.bar(results['Algorithm'], results['Accuracy'])
plt.xticks(rotation=45)
plt.ylabel('Accuracy')
plt.show()
```

```
from sklearn.metrics import confusion_matrix, classification_report
print(confusion_matrix(y_test, y_pred3))
print(classification_report(y_test, y_pred3))
```
